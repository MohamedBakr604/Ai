```python

# Initial skills dictionary
skills = [
    "python",
    "java",
    "javascript",
    "typescript",
    "dart",
    "flutter",
    "react",
    "angular",
    "vue",
    "node.js",
    "express",
    "django",
    "flask",
    "fastapi",
    "spring boot",
    "c#",
    ".net",
    "asp.net",
    "php",
    "laravel",
    "html",
    "css",
    "bootstrap",
    "tailwind",
    "sql",
    "mysql",
    "postgresql",
    "mongodb",
    "firebase",
    "machine learning",
    "deep learning",
    "artificial intelligence",
    "data analysis",
    "data science",
    "pandas",
    "numpy",
    "scikit-learn",
    "tensorflow",
    "pytorch",
    "computer vision",
    "nlp",
    "git",
    "github",
    "docker",
    "aws",
    "azure",
    "figma",
    "ui/ux",
    "rest api",
    "api integration"
]

print("Total Skills:", len(skills))
print(skills)
```

```python
import re

def extract_skills(text):
    found_skills = []

    for skill in skills:
        pattern = r"(?<!\w)" + re.escape(skill) + r"(?!\w)"

        if re.search(pattern, text):
            found_skills.append(skill)

    return found_skills
```

```python
# Extract skills from project title and description
projects["required_skills"] = (
    projects["title_clean"] + " " +
    projects["description_clean"]
).apply(extract_skills)
```

```python
# Display sample projects with extracted skills
display(
    projects[
        ["title", "required_skills"]
    ].head(20)
)
```

```python
# Check extraction results
print("Projects with Skills:",
      (projects["required_skills"].str.len() > 0).sum())

print("Projects without Skills:",
      (projects["required_skills"].str.len() == 0).sum())
```

```python
from collections import Counter

# Count all extracted skills
all_skills = []

for project_skills in projects["required_skills"]:
    all_skills.extend(project_skills)

skill_counts = Counter(all_skills)
```

```python
# Display top 30 skills
top_skills = pd.DataFrame(
    skill_counts.most_common(30),
    columns=["Skill", "Project_Count"]
)

display(top_skills)

print("Total Unique Skills:", len(skill_counts))
```
